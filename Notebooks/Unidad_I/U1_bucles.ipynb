# <font color="green"> Estructuras Iterativas (ciclos)

Una estructura iterativa es un tipo de estructura de control que permite controlar el flujo de datos, ejecutando, de manera eficiente y automatizada, una misma acción de forma repetida (*n veces*) o hasta que una condición se cumpla. A este tipo de estructuras también se les denomina reiterativa, cíclica, repetitiva o bucle (loop). 

### <font color="green">**¿Para qué nos sirven?**

- Procesar cada elemento de un conjunto de datos (temperaturas, mediciones, nombres de estaciones…).  
- Automatizar cálculos o acciones repetitivas (convertir unidades, etiquetar datos, generar reportes).  
- Implementar lógica de “hasta que ocurra X” (por ejemplo, pedir input al usuario hasta que sea válido).

### Bucle `for`

Itera sobre los elementos de cualquier **secuencia** u **objeto iterable** (listas, cadenas, tuplas, diccionarios, `range`, etc.).

```python
for <variable> in <iterable>:
    # bloque que se repite para cada elemento
```

El `for` de Python **no usa un contador manual**: simplemente recorre cada elemento uno a uno.

````python
# For sobre una lista
frutas = ["manzana", "pera", "uva", "mango"]
print("Frutas:")
for fruta in frutas:
    print(f"{fruta}")
````

````python
# enumerate() — índice + valor
# A veces necesitamos el índice mientras iteramos.
# IMPORTANTE: enumerate() nos da tuplas (índice, valor).

planetas = ["Mercurio", "Venus", "Tierra", "Marte", "Júpiter"]
print("Planetas del sistema solar:")
for i, planeta in enumerate(planetas, start=1):  # start=1, empieza en 1
    print(f"  {i}. {planeta}")
````

````python
# for sobre diccionarios

inventario = {
    #clave : valor,
    "manzana": 50,
    "naranja": 30,
    "pera":    20,
    "uva":    100,
}

# Solo claves (por defecto)
print("Claves:")
for clave in inventario:
    print(f"  {clave}")
````

````python
# Claves y valores con .items()
print("\nInventario completo:")
for producto, cantidad in inventario.items():
    estado = "STOCK" if cantidad >= 25 else "BAJO STOCK"
    print(f"  {producto:<10}: {cantidad:3d} unidades  {estado}")
````

````python

# Solo valores con .values()
total = 0
for cantidad in inventario.values():
    total += cantidad
print(f"\nTotal en inventario: {total} unidades")
````

````python
# Podemos mezclar el uso de if y for: Ejemplo aplicando filtros...

numeros = [3, 17, 42, 8, 99, 56, 23, 74, 11, 60]
pares    = []
impares  = []
mayores  = []

for n in numeros:
    if n % 2 == 0:
        pares.append(n)
    else:
        impares.append(n)
    if n > 50:
        mayores.append(n)

print(f"Números:  {numeros}")
print(f"Pares:    {pares}")
print(f"Impares:  {impares}")
print(f"Mayores de 50: {mayores}")
print(f"Suma total: {sum(numeros)}")
print(f"Promedio:   {sum(numeros)/len(numeros):.2f}")
````

## Bucle `while`

Repite un bloque **mientras** una condición sea verdadera. Se usa cuando **no sabemos de antemano** cuántas veces necesitamos repetir.

```python
while <condición>:
    # bloque que se repite mientras condición sea True
```

**Cuidado con los bucles infinitos**: siempre asegúrate de que la condición pueda volverse `False` en algún momento.

````python
# Ejemplo 1.
contador = 1
while contador <= 5:
    print(f"  Vuelta {contador}")
    contador += 1   #CRUCIAL: si olvidamos este paso, se vuelve un bucle infinito
print("¡Terminó el while!")
````

````python
# Ejemplo 3. Calcular dígitos de un número
numero = 123456789
digitos = 0
temp = numero

while temp > 0:
    digitos += 1
    temp //= 10    # Eliminar el último dígito
print(f"El número {numero} tiene {digitos} dígitos")
````

````python
# Ejemplo: buscar con while
# Buscar la primera potencia de 2 que supere 1000
potencia = 1
exponente = 0

while potencia <= 1000:
    exponente += 1
    potencia = 2 ** exponente
print(f"\nLa primera potencia de 2 que supera 1000 es:")
print(f"  2^{exponente} = {potencia}")
````

### ¿Cuándo usar for vs while?
Usa FOR cuando:
   - Conoces de antemano el número de iteraciones
   - Quieres recorrer una colección

 Usa WHILE cuando:
   - No sabes cuántas veces se repetirá
   - La repetición depende de una condición externa


Ejemplo: mismo resultado, distinta herramienta
```
print("Con FOR (número fijo de iteraciones):")
for i in range(1, 6):
    print(f"  i = {i}")

print("\nCon WHILE (condición variable):")
i = 1
while i <= 5:
    print(f"  i = {i}")
    i += 1
```

### **`break`, `continue`, `pass`**

Estas tres palabras se usan dentro de ciclos (for y while) y controlan el flujo del programa.

* `break`:Una instrucción que termina el ciclo inmediatamente, sin importar si la condición sigue siendo verdadera.
* `continue`:Salta el resto del código de esa iteración y pasa a la siguiente repetición.
* `pass`: Instrucción que no hace nada. Se usa cuando Python exige una instrucción, pero aún no quieres escribir código.
----
### **Ejemplos con`break`**

````python
#Ejemplo 1
for i in range(10):
    if i == 4:
        break
    print(i)  #Sale del ciclo en el momento en que se ejecuta.
````

````python
#Ejemplo 2
for i in range(10):
    print(i)
    if i == 4:
        break
    
````

````python
#Ejemplo 3

while True: #Empezamos diciendo que siempre será verdadero
    numero = int(input("Escribe 0 para salir: "))
    if numero == 0: #Mientras no se cumpla la condición el ciclo sigue
        break
````

````python
#Ejemplo 4
alturas_ola = [0.5, 0.8, 1.2, 3.5, 0.9]

for altura in alturas_ola:
    if altura >= 3:
        print("Posible tsunami detectado")
        break
    print("Altura normal:", altura)

````

### **Ejemplos con `continue`**

````python
#Ejemplo 1

for i in range(5):
    if i == 2:
        continue  #No termina el ciclo, solo omite lo que sigue en esa vuelta.
    print(i)
````

````python
#Ejemplo 2

numero = 0

while numero < 5:
    numero += 1
    
    if numero == 3:
        continue
    
    print(numero)
````

````python
#Ejemplo 3

salinidad = [35, 36, None, 34, 37]

for valor in salinidad:
    if valor is None:
        continue
    print("Salinidad válida:", valor)
````

### Ejemplos con `pass`

Permite dejar preparada la estructura sin romper el programa.

````python
#Ejemplo 1

for i in range(5):
    if i == 2:
        pass
    print(i)
````

````python
#Ejemplo 2

for estacion in ["seca", "lluviosa"]:
    if estacion == "seca":
        pass  # Se agregará análisis de sequía después
    else:
        print("Analizando temporada lluviosa")
````
