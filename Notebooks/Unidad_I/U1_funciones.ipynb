---

## ¿Qué son las funciones?

Una función es un **bloque de código reutilizable**:  
- Se puede ejecutar muchas veces dentro de un programa.  
- Recibe datos de entrada (*argumentos*).  
- Procesa esos datos.  
- Devuelve un resultado de salida.  

---

## ¿Por qué usar funciones?

Entre las principales ventajas están:

- **Dividir y ordenar el código** en partes más sencillas → facilita depuración y programación.  
- Reutilizar código, evitando repeticiones innecesarias.  
- Cumplir con el principio de no repetir código DRY (*Don't Repeat Yourself*).  

---

## Tipos de funciones en Python

1. **Funciones nativas** (*built-in functions*): ya vienen integradas en Python.  
   Ejemplos:  
   - `print()`  
   - `range()`  
   - `list()`  

2. **Funciones personalizadas** (*user-defined functions*): creadas por el usuario.  
   → En este curso nos vamos a enfocar principalmente en este tipo.

````python
# Juguemos un poco con las cademas, por ejemplo, podemos hacer cambios, buscar, reemplazar, etc.
texto = "  Hola, Mundo! Esto es Python.  " #Noten que tenemos letras (mayúsculas y minúsculas), signos, espacios, etc.

# Mayúsculas / Minúsculas
print(texto.upper())        # Todo a mayúsculas
print(texto.lower())        # Todo a minúsculas
print(texto.capitalize())   # Solo primera letra en mayúscula
print(texto.title())        # Primera letra de cada palabra en mayúscula
print(texto.swapcase())     # Invierte mayúsculas/minúsculas

print("─" * 40)
````

````python
# Eliminar espacios en blanco
print(repr(texto.strip()))   # Elimina de ambos lados
print(repr(texto.lstrip()))  # Elimina del lado izquierdo
print(repr(texto.rstrip()))  # Elimina del lado derecho

print("─" * 40)
````

````python
# Búsqueda y reemplazo
base = "El sol brilla y el sol calienta."
print(base.find("sol"))               # Posición de la primera ocurrencia
print(base.count("sol"))              # Cuántas veces aparece
print(base.replace("sol", "solecito"))      # Reemplazar todas las ocurrencias
print(base.replace("sol", "astro", 1)) # Reemplazar solo la primera

print("─" * 40)
````

````python
# Verificaciones (retornan True o False)
print("1p345".isdigit())    # ¿Es todo dígitos?
print("Python3".isalpha())   # ¿Es todo alfabético?
print("Hola".islower())     # ¿Todo en minúsculas?
print("HOLA".isupper())     # ¿Todo en mayúsculas?
print("   ".isspace())      # ¿Todo son espacios?
print("Python".startswith("Py"))  # ¿Empieza con?
print("Python".endswith("on"))    # ¿Termina con?
````

````python
# Además podemos usar algunas funciones para manipular estas cadenas...

## Función split() — divide la cadena en una lista de palabras
frase = "Python es un lenguaje poderoso"
palabras = frase.split()           # Divide por espacios por defecto
print(f"split():    {palabras}")
````

````python
csv_data = "manzana,pera,naranja,uva"
frutas = csv_data.split(",")       # Divide por coma
print(f"split(','): {frutas}")
````

````python
## Función join() — une una lista de strings con un separador
unido = " | ".join(frutas)
print(f"join('|'):  {unido}")

print("─" * 40)
````

````python
## Concatenación
# Formas de formatear cadenas
nombre = "María"
nota   = 9.75

# 1. Concatenación con +  (no recomendado, no mezcla tipos)
print("Hola " + nombre)

# 3. str.format()
print("Hola {}, tu nota es {:.2f}".format(nombre, nota))

# 4. f-strings
print(f"Hola {nombre}, tu nota es {nota:.2f}")
print(f"El doble de la nota es: {nota * 2:.2f}")
````

### Librería math
El módulo math tiene funciones matemáticas avanzadas. Debes importarlo antes de usarlo.

````python
# IMPORTAR MÓDULOS

# Forma 1: importar el módulo completo
import math
print(math.pi) # Se accede con math.función()
````

````python
# CONSTANTES MATEMÁTICAS

print("Constantes:")
print(f"  math.pi  = {math.pi}")   # π ≈ 3.14159...
print(f"  math.e   = {math.e}")    # Número de Euler ≈ 2.71828...
print(f"  math.tau = {math.tau}")  # τ = 2π
print(f"  math.inf = {math.inf}")  # Infinito
print(f"  math.nan = {math.nan}")  # Not a Number
````

````python
# FUNCIONES DE REDONDEO Y VALOR ABSOLUTO
# Ejemplos
y = -2.3
print(f"y = {y}")
print(f"  math.floor(y) = {math.floor(y)}")  # -3 (no -2!)
print(f"  math.fabs(y)  = {math.fabs(y)}")   # Valor absoluto como float
````

````python
# POTENCIAS, RAÍCES Y LOGARITMOS

print("Raíces y potencias:")
print(f"  math.sqrt(144)   = {math.sqrt(144)}")    # Raíz cuadrada
print(f"  math.sqrt(2)     = {math.sqrt(2):.6f}")  # √2
print(f"  math.pow(2, 0.5) = {math.pow(2, 0.5):.6f}")  # También √2
print(f"  math.pow(3, 3)   = {math.pow(3, 3)}")    # 3^3 (retorna float)
print(f"  math.isqrt(17)   = {math.isqrt(17)}")    # Raíz cuadrada ENTERA
````

````python


print("\nLogaritmos:")
print(f"  math.log(math.e) = {math.log(math.e)}")      # Logaritmo natural (base e)
print(f"  math.log(100)    = {math.log(100):.4f}")     # ln(100)
print(f"  math.log(100,10) = {math.log(100, 10)}")     # log base 10
print(f"  math.log10(1000) = {math.log10(1000)}")      # log10 (más preciso)
print(f"  math.log2(1024)  = {math.log2(1024)}")       # log base 2
````

````python
print("\nExponencial:")
print(f"  math.exp(1)  = {math.exp(1):.6f}")   # e^1 = e
print(f"  math.exp(0)  = {math.exp(0)}")       # e^0 = 1
print(f"  math.exp(3)  = {math.exp(3):.6f}")   # e^3
````

````python
# FUNCIONES TRIGONOMÉTRICAS
# NOTA: Todas trabajan en RADIANES, no en grados.

# Conversión entre grados y radianes
grados   = 90
radianes = math.radians(grados)      # Grados → Radianes
de_vuelta = math.degrees(radianes)   # Radianes → Grados

print(f"90 grados = {radianes:.6f} radianes")
print(f"{radianes:.6f} radianes = {de_vuelta:.1f} grados")
````

````python
print("\nFunciones trigonométricas (en grados → radianes):")
for grado in [0, 30, 45, 60, 90, 180]:
    rad  = math.radians(grado)
    seno = math.sin(rad)
    cose = math.cos(rad)
    # math.tan(90°) tiende a infinito, lo manejamos
    tang = math.tan(rad) if grado != 90 else float('inf')
    print(f"  {grado:3}°: sin={seno:+.4f}, cos={cose:+.4f}, tan={tang:+.4f}")
````

````python
# Funciones inversas
print("\nFunciones inversas:")
print(f"  asin(1.0) = {math.degrees(math.asin(1.0)):.1f}°")
print(f"  acos(0.0) = {math.degrees(math.acos(0.0)):.1f}°")
print(f"  atan(1.0) = {math.degrees(math.atan(1.0)):.1f}°")
````

````python
# OTRAS FUNCIONES ÚTILES

# factorial()
print("Factoriales:")
for n in range(8):
    print(f"  {n}! = {math.factorial(n)}")
````

````python
# gcd() — Máximo Común Divisor
print(f"\ngcd(48, 36)  = {math.gcd(48, 36)}")
print(f"gcd(100, 75) = {math.gcd(100, 75)}")
````

````python

# lcm() — Mínimo Común Múltiplo (Python 3.9+)
print(f"lcm(4, 6)    = {math.lcm(4, 6)}")
print(f"lcm(12, 18)  = {math.lcm(12, 18)}")
````

````python
# isnan() e isinf() — comprobar valores especiales
print(f"\nisinf(math.inf) = {math.isinf(math.inf)}")
print(f"isnan(math.nan) = {math.isnan(math.nan)}")
print(f"isfinite(3.14)  = {math.isfinite(3.14)}")
````

````python
# dist() — distancia euclídea entre dos puntos (Python 3.8+)
p1 = (0, 0)
p2 = (3, 4)
print(f"\nDistancia de {p1} a {p2} = {math.dist(p1, p2)}")  # √(9+16) = 5.0
````

### `return` vs `print`

|`print`|`return`|
|--------|-------|
|Solo muestra información en pantalla|Devuelve un valor|

## Argumentos en funciones

Hasta ahora hemos usado sólo un argumento en las funciones,

- **Obligatorios**: siempre se deben proporcionar.

## Formas de pasar argumentos

1. **Posicional**  
   Los argumentos se agregan en el mismo orden en que aparecen los parámetros en la definición de la función.  

   ```python

       print(f"Hola, soy {nombre} y estudio {carrera}")

   # Uso posicional (orden importa)
   presentacion("María", "Ciencias de la Tierra")
   ```

#### Recordatorio
El **factorial** de un número entero positivo `n` se define como:  

$n! = n \times (n-1) \times (n-2) \times \dots \times 1 $

Ejemplo:  
$
5! = 5 \times 4 \times 3 \times 2 \times 1 = 120$
