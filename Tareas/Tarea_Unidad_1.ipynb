# <font color="purple">**Ejercicio 1: Monitoreando la temporada de huracanes** </font>

Durante la temporada de huracanes, el monitoreo de los sistemas tropicales es fundamental para conocer su evolución y evaluar los posibles riesgos asociados. Estos riesgos van desde lluvias intensas y vientos dañinos hasta marejadas e inundaciones, aspectos que dependen directamente de la intensidad y el tipo de ciclón al que se enfrenta la población.

Por ello, en este ejercicio desarrollarán un **programa que permita registrar los datos de 10 sistemas tropicales y clasificarlos automáticamente de acuerdo con su velocidad máxima del viento**.

1. Pide al usuario, para cada uno de los 10 sistemas tropicales, que ingrese: *región oceánica donde se formó (por ejemplo, Pacífico u Atlántico)*, *velocidad máxima sostenida del viento en km/h* y *presión atmosférica mínima en hPa*.
2. Clasifica el sistema tropical guiándote de la siguiente tabla:

      |Categoría|Vientos máximos sostenidos (km/h)|
      |:-------:|:---:|
      |Depresión tropical| < 63|
      |Tormenta tropical|63-118|
      |Huracán categoría 1|119-153|
      |Huracán categoría 2|154-177|
      |Huracán categoría 3|178-208|
      |Huracán categoría 4|209-251|
      |Huracán categoría 5|≥ 252|

3. Almacena la información de los sistemas en un DataFrame que contenga la región oceánica de formación, la velocidad máxima del viento, la presión mínima y la categoría. Cada vez que el usuario ingrese los datos de un sistema tropical, actualiza el DataFrame incorporando el nuevo registro.

4. Agrupa los sistemas de acuerdo con la región donde se formaron y obtenga para cada región:

- El promedio de la velocidad máxima del viento.
- El promedio de la presión mínima.
- El número de sistemas registrados.

5. Imprime los resultados del paso 4 en pantalla.

**Extra (+5 puntos) Modifica tu programa de manera que se encuentre dentro de una función.**

# <font color="purple"> **Ejercicio 4: Clasificador de Rocas** </font>

Crea un programa que permita al usuario ingresar datos de rocas (densidad, textura y dureza), almacenarlos en listas, y clasificarlos según criterios simples. El usuario podrá consultar las rocas ingresadas y ver su clasificación. Piensa este ejercicio como una herramienta para las prácticas de campo: Cada que tomas una muestra ingresas sus datos "al sistema", los clasificas automáticamente y los guardas para verlos después.

Si lo prefieres, crea un clasificador de nubes, huracánes u otra cosa que sea su interés (siempre y cuando usen al menos 3 criterios para la clasificación).

### Instrucciones

1. Crea tres listas vacías: `densidades`, `dureza` , `texuras`.
2. Crea una función  que retorne una clasificación:

    - Ígnea: densidad > 2.8 y textura = "granular"
    - Sedimentaria: densidad < 2.6 y textura = "suave"
    - Metamórfica: dureza > 7 o textura = "laminar"
    - Desconocida: si no cumple ninguna condición
3. Usa un bucle  para permitir al usuario ingresar múltiples rocas.
4. Al final, muestra todas las rocas ingresadas con su clasificación.


Extra (+5 puntos): Conviertan las listas a un arreglo de Numpy o en un DataFrame de Pandas.

Como sismólogo, necesitas analizar datos de terremotos registrados en diferentes estaciones sísmicas de México. Tu tarea es organizar y procesar esta información usando diccionarios y listas.

#### Instrucciones

1. Crea un diccionario llamado `estaciones` donde cada clave sea el código de una estación (ej: "CDMX01", "GDL02", "MTY03") y cada valor contenga:

5. Extrae todas las magnitudes de los eventos y crea un arreglo de NumPy.
6. Calcula la magnitud promedio, máxima y mínima.
7. Determina cuántos sismos superan magnitud 4.0 (considerados moderados).

8. Crea una lista con los IDs de sismos que ocurrieron a menos de 10 km de profundidad (sismos superficiales).
